# 04 — Exploratory data analysis
Explore order-level trends and distributions after validating the source schema.

```python
import sys
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from ecommerce_analytics.config import ConfigManager
from ecommerce_analytics.analysis import EDAAnalyzer
from ecommerce_analytics.visualization import EDAVisualizer

# Instantiate analysis and visual engines
analyzer = EDAAnalyzer()
visualizer = EDAVisualizer()
```

```python
# Aggregate monthly sales performance
monthly_df = analyzer.get_monthly_revenue_trend()
display(monthly_df.tail(10))

# Export plot and show Inline
chart_path = visualizer.plot_monthly_revenue_trend(monthly_df)
print(f"Saved figure to: {chart_path}")

# Display in Notebook
plt.figure(figsize=(10, 4))
df_display = monthly_df[monthly_df["purchase_year_month"] < "2018-09"]
sns.lineplot(data=df_display, x="purchase_year_month", y="total_revenue", marker="o", color="#1f77b4")
plt.title("Monthly Gross Revenue (GMV)")
plt.xticks(rotation=45)
plt.ylabel("Revenue (BRL)")
plt.tight_layout()
plt.show()
```

```python
# Compute top categories
top_cats = analyzer.get_top_categories_by_revenue(10)
display(top_cats)

# Render figure inline
plt.figure(figsize=(9, 5))
sns.barplot(data=top_cats, x="total_revenue", y="product_category_name_english", palette="Blues_r")
plt.title("Top 10 Grossing Categories")
plt.xlabel("Total Sales Revenue (BRL)")
plt.tight_layout()
plt.show()
```
