# 03 — Data cleaning
Apply conservative, configurable cleanup and save results under `data/interim/`. Raw files remain unchanged.

```python
from pathlib import Path
import sys
project_root = Path.cwd()
if not (project_root / "src").exists():
    project_root = project_root.parent
sys.path.insert(0, str(project_root / "src"))
from ecommerce_analytics.config import load_config
from ecommerce_analytics.data_loader import load_tables, save_tables
from ecommerce_analytics.cleaning import clean_table

config = load_config()
tables = load_tables(config)
cleaned = {name: clean_table(table, config["cleaning"]) for name, table in tables.items()}
save_tables(cleaned, Path(config["paths"]["interim"]))
{name: table.shape for name, table in cleaned.items()}
```
